```python
# How do we represent the layers?
# How do we keep track of the no. of neurons in each layer?
# How do we represent the weights?

def count_params(S):
  L = len(S)
  params = 0
  for l in range(1,L):
    #W_l
    params += (S[l]*S[l-1])
    #b_l
    params += S[l]
  return params

S = [10,100,3] # 10 - Input neurons, 100- Hidden layers, 3 - output neurons
count_params(S)
```
